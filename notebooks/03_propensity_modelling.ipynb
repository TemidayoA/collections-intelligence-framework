# 03: Propensity-to-Pay Modelling

This notebook builds and evaluates propensity-to-pay models.

## Key Insight

The most powerful application of propensity models is not deciding **who to contact**, but **who NOT to contact**.

Suppressing unnecessary contact:
- Reduces operational cost
- Prevents borrower fatigue
- Preserves goodwill
- Does not sacrifice recovery (high self-cure accounts pay anyway)
