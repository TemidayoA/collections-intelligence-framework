## 1. Metrics That Matter

Shift from activity metrics to outcome metrics.

```python
# Old metrics (to deprecate)
old_metrics = {
    'Calls per Agent': 'Rewards activity, not outcomes',
    'Messages Sent': 'Encourages spam',
    'Contact Attempts': 'Volume ≠ effectiveness'
}

# New metrics (to adopt)
new_metrics = {
    'Recovery per Contact': 'Collections / Contact Attempts',
    'Contact Efficiency Ratio': 'Successful Outcomes / Total Contacts',
    'Self-Cure Rate': 'Accounts recovered without contact',
    'Customer Reactivation': 'Repeat borrowing after delinquency'
}

print("Metrics to DEPRECATE:")
for metric, problem in old_metrics.items():
    print(f"  ❌ {metric}: {problem}")

print("\nMetrics to ADOPT:")
for metric, definition in new_metrics.items():
    print(f"  ✓ {metric}: {definition}")
```
