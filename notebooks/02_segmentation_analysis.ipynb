# 02: Segmentation Analysis

This notebook implements and validates the Willingness-Capacity Matrix segmentation framework.

## The Framework

Traditional collections segment borrowers as 'good' or 'bad'. This binary view undermines effectiveness.

The Willingness-Capacity Matrix segments on two dimensions:
- **Willingness**: Intent to repay (behavioural signals)
- **Capacity**: Ability to repay (financial signals)

This produces four segments with distinct optimal strategies.

## 1. Feature Preparation

Prepare willingness and capacity indicators.

```python
# Willingness indicators
willingness_features = [
    'payment_attempts_30d',      # Failed payment attempts = strong intent
    'contact_response_rate',     # Responsive = engaged
    'proactive_contact',         # Reached out = high intent
    'ptp_kept_ratio',            # Kept promises = reliable
    'channels_blocked'           # Blocking = avoidance (negative)
]

# Capacity indicators
capacity_features = [
    'income_regularity_score',   # Regular income = stable capacity
    'transaction_activity_score', # Active = financially healthy
    'previous_completion_rate',   # Track record
    'debt_to_income_ratio',       # Over-leveraged = low capacity (negative)
    'account_age_months'          # Tenure = stability
]
```

## 2. Apply Segmentation

```python
# Initialize matrix
matrix = WillingnessCapacityMatrix(
    willingness_threshold=0.5,
    capacity_threshold=0.5
)

# Example segmentation
sample_features = {
    'payment_attempts_30d': 2,
    'contact_response_rate': 0.7,
    'proactive_contact': True,
    'ptp_kept_ratio': 0.6,
    'channels_blocked': 0,
    'income_regularity_score': 0.8,
    'transaction_activity_score': 0.6,
    'previous_completion_rate': 1.0,
    'debt_to_income_ratio': 0.3,
    'account_age_months': 12
}

result = matrix.segment('EXAMPLE_001', sample_features)
print(f"Segment: {result.segment.value}")
print(f"Willingness: {result.willingness_score:.2f}")
print(f"Capacity: {result.capacity_score:.2f}")
print(f"Action: {result.recommended_action}")
```
