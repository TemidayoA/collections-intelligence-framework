## 3. Temporary vs Structural Default

Key insight: Not all delinquency is the same.

- **Temporary delinquency**: Borrower intends to pay but faces short-term liquidity constraints
- **Structural default**: Borrower cannot or will not repay

Identifying which type a borrower exhibits is crucial for intervention strategy.

```python
# Indicators of temporary delinquency:
# - Payment attempts (even if failed)
# - Responsive to contact
# - Regular income pattern detected
# - DPD stabilising (not accelerating)

# Indicators of structural default:
# - No payment attempts
# - Avoiding contact (channel blocking)
# - Multiple concurrent delinquencies
# - DPD accelerating rapidly
```
